```python
import scipy.io
import numpy as np
matCentral = scipy.io.loadmat('detector_efficiency_central_SINQ_SANS_I_HDF.mat')
```

```python
maskx,masky =np.where(matCentral['eff_data']==1)
print()
```

```python
mateff=np.copy(matCentral['eff_data'])
effx=np.sum(mateff,where=(mateff!=1),axis=0,)/np.sum(mateff!=1,axis=0)
effy=np.sum(mateff,where=(mateff!=1),axis=1,)/np.sum(mateff!=1,axis=1)
effg = np.empty([128, 128])
for i in range(len(effx)):
    for j in range(len(effy)):
        effg[j,i] = effx[i]*effy[j]
import matplotlib.pyplot as plt
plt.imshow(matCentral['eff_data'])
plt.show()
```

```python
effnew = np.copy(mateff)
effscaling = np.average(effg)
print(effscaling)
effnew[mateff==1] = effg[mateff==1]/effscaling
```

```python
import matplotlib.pyplot as plt
fig = plt.figure()
ax1 = fig.add_subplot(1,3,1)
ax1.imshow(matCentral['eff_data'],vmin=0,vmax=1.4)
ax2 = fig.add_subplot(1,3,2)
ax2.imshow(effg/np.average(effg),vmin=0,vmax=1.4)
ax3 = fig.add_subplot(1,3,3)
ax3.imshow(effnew,vmin=0,vmax=1.4)
plt.show()
```

```python
matCentral['eff_data']=effnew
scipy.io.savemat('detector_efficiency_extrapolate_SINQ_SANS_I_HDF.mat',matCentral)
```
